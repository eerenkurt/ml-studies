```python
from sklearn.linear_model import LinearRegression
import pandas as pd
```

```python
# y = mx + b (y ler dependent, x ler independent) (x ler y yi etkiler)
```

```python
df=pd.read_csv("Student_Marks.csv")
```

```python
df.head(3)
```

```python
y = df[["Marks"]]
x = df[["number_courses" , "time_study"]]
```

```python
df.info()
```

```python
# sklearn.linear_model.LinearRegression() (yukarda çağırmasaydık)
lin = LinearRegression()
model = lin.fit(x,y)
```

```python
model.predict([[4,10]])
```

```python
df["Marks"].max()
```

```python
model.score(x,y)
```
