```python
from sklearn.linear_model import LinearRegression
import pandas as pd
```

```python
df = pd.read_csv("student.csv")
```

```python
df.head(3)
```

```python
df.columns = ["sinif", "saat", "puan"]
```

```python
df.head(3)
```

```python
y = df[["puan"]]
x = df[["sinif", "saat"]]
```

```python
lin = LinearRegression()
model = lin.fit(x,y)
```

```python
# y = ax + b
```

```python
model.coef_ #katsayı
```

```python
model.intercept_ #sabit (b olan kısım yani)
```

```python
model.predict([[4,5]])
```

```python
df.head(3)
```

```python
s1 = model.predict([[3,4.508]])[0][0]
```

```python
(19.202-s1)/19.202
```

```python
model.score(x,y)
```
